### Step 1. Import the necessary libraries:

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv).

### Step 3. Assign it to a variable called tips

```python
url = 'https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv'
tips = pd.read_csv(url);
tips.head()
```

### Step 4. Delete the Unnamed 0 column

```python
tips.drop('Unnamed: 0', axis=1, inplace=True)
tips.head()
```

### Step 5. Plot the total_bill column histogram

```python
# create histogram
ttbill = sns.histplot(tips.total_bill);

# set lables and titles
ttbill.set(xlabel = 'Value', ylabel = 'Frequency', title = "Total Bill")

# take out the right and upper borders
sns.despine()
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
plt.scatter(tips.total_bill, tips.tip)
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Total Bill vs Tip')
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.
#### Hint: It is just one function.

```python
plt.scatter(tips.total_bill, tips.tip, s=tips.size*0.1)
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.title('Total Bill vs Tip')
```

### Step 8. Present the relationship between days and total_bill value

```python
sns.boxplot(x='day', y='total_bill', data=tips)
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.title('Total Bill by Day')
```

### Step 9. Create a scatter plot with the day as the y-axis and tip as the x-axis, differ the dots by sex

```python
sns.scatterplot(x='tip', y='day', hue='sex', data=tips)
plt.xlabel('Tip')
plt.ylabel('Day')
plt.title('Tip vs Day, Differentiated by Sex')
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
sns.boxplot(x='day', y='total_bill', hue='time', data=tips)
plt.xlabel('Day')
plt.ylabel('Total Bill')
plt.title('Total Bill by Day, Differentiated by Time')
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(tips[tips['time'] == 'Dinner']['tip'], ax=ax[0], kde=True)
ax[0].set_title('Tip Distribution for Dinner')
ax[0].set_xlabel('Tip Amount')
ax[0].set_ylabel('Frequency')

sns.histplot(tips[tips['time'] == 'Lunch']['tip'], ax=ax[1], kde=True)
ax[1].set_title('Tip Distribution for Lunch')
ax[1].set_xlabel('Tip Amount')
ax[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 6))

# Scatterplot for Male customers
sns.scatterplot(x='total_bill', y='tip', hue='smoker', data=tips[tips['sex'] == 'Male'], ax=ax[0])
ax[0].set_title('Total Bill vs Tip for Male Customers')
ax[0].set_xlabel('Total Bill ($)')
ax[0].set_ylabel('Tip ($)')

# Scatterplot for Female customers
sns.scatterplot(x='total_bill', y='tip', hue='smoker', data=tips[tips['sex'] == 'Female'], ax=ax[1])
ax[1].set_title('Total Bill vs Tip for Female Customers')
ax[1].set_xlabel('Total Bill ($)')
ax[1].set_ylabel('Tip ($)')

plt.tight_layout()
plt.show()
```
