1. Import the necessary libraries

```python
import pandas as pd
```

 Step 2. Import the dataset from this address.

```python
worldcup = pd.read_csv("/content/WorldCupMatches.csv")
```

Step 3. Count the total number of matches played in each World Cup year.

```python
worldcup.groupby('Year')['Year'].count()
```

Step 4: Calculate the total attendance for each World Cup year.

```python
worldcup.groupby('Year')['Attendance'].sum()
```

Step 5: Find the average number of goals scored by the home team in each World Cup year.

```python
worldcup.groupby('Year')['Home Team Goals'].mean()
```

Step 6: Find the average number of goals scored by the away team in each World Cup year.

```python
worldcup.groupby('Year')['Away Team Goals'].mean()
```

Step 7: Count the number of matches played in each tournament stage.

```python
worldcup.groupby('Stage')['Stage'].count()
```

Step 8: Count how many matches were played in each host city.

```python
worldcup.groupby('City')['City'].count()
```

Step 9: Count the number of matches played in each stadium.

```python
worldcup.groupby('Stadium')['Stadium'].count()
```

Step 10. Find the total number of matches officiated by each referee

```python
worldcup.groupby('Referee')['Referee'].count()
```

Step 11. Find the number of matches played by each home team.

```python
worldcup.groupby('Home Team Name')['Home Team Name'].count()
```

Step 12. Find the number of matches played by each away team.

```python
worldcup.groupby('Away Team Name')['Away Team Name'].count()
```
